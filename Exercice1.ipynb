```python
import numpy as np
np.random.seed(42)

import numpy as np

np.random.seed(42)

def generate_case2_data(n):
    X = np.random.normal(loc=0.0, scale=1.0, size=n)
    U = np.random.uniform(low=0.0, high=1.0, size=n)
    # Conditional probabilities: P(Y=1 | X <= 0) = 0.2, P(Y=1 | X > 0) = 0.6
    Y = np.where(X <= 0, (U <= 0.2).astype(int), (U > 0.4).astype(int))
    return X.reshape(-1, 1), Y

# 1.1 Training sample (n = 1000)
X_train, y_train = generate_case2_data(1000)

# 1.2 Test sample (n = 200)
X_test, y_test = generate_case2_data(200)

```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

```python


# Fit classifier on training set
knn = KNeighborsClassifier(n_neighbors=9)
knn.fit(X_train, y_train)

# Predictions
y_pred_train = knn.predict(X_train)
y_pred_test = knn.predict(X_test)

# Error computation (Misclassification Rate = 1 - Accuracy)
train_error = 1.0 - accuracy_score(y_train, y_pred_train)
test_error = 1.0 - accuracy_score(y_test, y_pred_test)

print(f"Training Misclassification Error: {train_error:.4f}")
print(f"Test Misclassification Error:     {test_error:.4f}")
```

```python
train_acc = accuracy_score(y_train, y_pred_train)
test_acc = accuracy_score(y_test, y_pred_test)
print(train_acc,test_acc)
```
